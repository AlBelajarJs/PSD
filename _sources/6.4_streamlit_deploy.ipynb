# 6.4 Sistem Informasi Streamlit + Deploy

## Menjalankan aplikasi

```bash
pip install -r requirements.txt
streamlit run streamlit_jatim.py
```

Buka `http://localhost:8501`. Dua mode:

1. **Tebak kelas** — isi 6 nilai reflektansi band (B02, B03, B04, B08, B11, B12, skala 0–10000) → model `jatim_best.joblib` (**XGBoost**, akurasi test 63.8%, F1-macro 0.6265) menghitung 6 indeks + memprediksi 1 dari 6 kelas + probabilitas.
2. **Klasifikasi CSV** — upload CSV berisi kolom B02..B12 → seluruh baris diprediksi + unduh hasil.

## Deploy

- **Web statis (dokumentasi proses + peta):** Jupyter Book ini → GitHub Pages (branch `gh-pages`, via `ghp-import`). Peta Folium interaktif: `jatim/folium_jatim_samples.html` (sebaran 2.823 sampel) dan `jatim/folium_demo_klasifikasi.html` (hasil klasifikasi + Google Satellite).
- **Aplikasi interaktif:** Streamlit Community Cloud (hubungkan repo → `streamlit_jatim.py`) atau server sendiri (`streamlit run --server.port 80`).

Peta layar-penuh (folder `maps/` di situs ini):

- [Sebaran 2.823 sampel Jatim](maps/folium_jatim_samples.html)
- [Hasil klasifikasi demo Sidoarjo–Surabaya](maps/folium_demo_klasifikasi.html)
- [Hasil klasifikasi Wonoayu sawah vs non-sawah](maps/folium_wonoayu.html)

## Metodologi & batas jujur

- **Data latih**: ekstraksi **per piksel** (30 titik acak/poligon, 84010
  sampel bersih) dari komposit median Sentinel-2A **ber-masking awan SCL**,
  bukan rata-rata per poligon — sehingga 100% poligon terwakili dan praktis tanpa missing value.
- **Evaluasi jujur**: split 80/20 **berbasis grup poligon** + `StratifiedGroupKFold`
  sehingga tidak ada piksel poligon yang bocor train<->test. Model terbaik
  **XGBoost**: akurasi test 63.8%, F1-macro 0.6265, gap CV-test 0.0161
  (kecil -> tidak overfitting).
- **Sisa kesalahan**: Sawah <-> Lahan hijau <-> Bangunan masih sering tertukar
  (kelas vegetasi spektralnya mirip). Naik butuh fitur tekstur/temporal.
- **Kelas Danau** paling sedikit (~9,4 ribu piksel); proksi Waduk.
- **Peta wall-to-wall** hanya jendela demo Sidoarjo-Surabaya @20 m, bukan full Jatim
  (full provinsi @10 m ≈ 6 GB, di luar lingkup UTS).
